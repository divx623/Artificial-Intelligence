```python
import math

def alpha_beta(depth, index, is_max, values, alpha, beta):
    indent = "  " * depth
    node_type = "MAX" if is_max else "MIN"

    # Level 3: Terminal leaf nodes
    if depth == 3:
        val = values[index]
        print(f"{indent}Depth {depth} [Terminal Node idx={index}]: Value = {val}")
        return val

    print(f"{indent}Depth {depth} [{node_type} Node idx={index}]: alpha={alpha}, beta={beta}")

    if is_max:
        best = -math.inf
        for i, direction in enumerate(["left", "right"]):
            child_idx = index * 2 + i
            print(f"{indent}  Evaluating {direction} child (idx={child_idx})...")
            val = alpha_beta(depth + 1, child_idx, False, values, alpha, beta)
            best = max(best, val)
            alpha = max(alpha, best)
            if beta <= alpha:
                print(f"{indent}  !! PRUNING !! v ({best}) >= beta ({beta}) at MAX level {depth} index {index}. Skipping remaining children.")
                break
        return best
    else:
        best = math.inf
        for i, direction in enumerate(["left", "right"]):
            child_idx = index * 2 + i
            print(f"{indent}  Evaluating {direction} child (idx={child_idx})...")
            val = alpha_beta(depth + 1, child_idx, True, values, alpha, beta)
            best = min(best, val)
            beta = min(beta, best)
            if beta <= alpha:
                print(f"{indent}  !! PRUNING !! v ({best}) <= alpha ({alpha}) at MIN level {depth} index {index}. Skipping remaining children.")
                break
        return best

# Given terminal leaf values
leaves = [10, 9, 14, 18, 5, 4, 50, 3]

print("=== STARTING ALPHA-BETA SEARCH ===")
optimal_value = alpha_beta(depth=0, index=0, is_max=True, values=leaves, alpha=-math.inf, beta=math.inf)
print(f"\nOptimal Minimax Value at Root: {optimal_value}")
```
