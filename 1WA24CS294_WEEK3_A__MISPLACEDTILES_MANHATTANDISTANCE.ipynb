```python
import heapq
def print_state(state):
    for row in state:
        print(row)
    print()

def misplaced_tiles(state, goal):
    count = 0
    for i in range(3):
        for j in range(3):
            # Ignore blank
            if state[i][j] != 0:
                if state[i][j] != goal[i][j]:
                    count += 1
    return count

def manhattan_distance(state, goal):
    total = 0
    for i in range(3):
        for j in range(3):
            tile = state[i][j]

            if tile != 0:
                for x in range(3):
                    for y in range(3):
                        if goal[x][y] == tile:
                            total += abs(i - x) + abs(j - y)
    return total


def generate_children(state):
    children = []

    state = [list(row) for row in state]
    blank_row, blank_col = 0, 0
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                blank_row = i
                blank_col = j

    moves = [
        (-1, 0), # Up
        (1, 0),  # Down
        (0, -1), # Left
        (0, 1)   # Right
    ]

    for dr, dc in moves:
        new_row = blank_row + dr
        new_col = blank_col + dc

        # Check if new position is inside the puzzle
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            # Make a copy
            new_state = [row[:] for row in state]
            # Move blank
            new_state[blank_row][blank_col] = new_state[new_row][new_col]
            new_state[new_row][new_col] = 0
            # Convert to tuple
            new_state = tuple(tuple(row) for row in new_state)
            children.append(new_state)

    return children

def A_STAR(start, goal, heuristic):
    OPEN = []
    CLOSED = set()
    g = {}
    parent = {}
    counter = 0

    start = tuple(tuple(row) for row in start)
    goal = tuple(tuple(row) for row in goal)

    g[start] = 0
    h = heuristic(start, goal)
    f = g[start] + h
    heapq.heappush(OPEN, (f, counter, start))
    parent[start] = None
    counter += 1

    print("\n=== STARTING A* SEARCH TRACE ===")
    print("Initial State:")
    print_state(start)
    print(f"Initial g=0, h={h}, f={f}\n")

    step_num = 1
    while OPEN:
        print(f"--- Step {step_num} ---")
        f_current, _, current = heapq.heappop(OPEN)

        if current in CLOSED:
            print(f"[REJECTED] Popped state already in CLOSED set. f-value: {f_current}")
            print_state(current)
            step_num += 1
            continue

        CLOSED.add(current)
        current_g = g[current]
        current_h = heuristic(current, goal)

        print(f"[VISITING] Active State (g={current_g}, h={current_h}, f={f_current}):")
        print_state(current)

        if current == goal:
            print("--- SUCCESS: Goal State Reached! ---")
            path = []
            temp = current
            while temp is not None:
                path.append(temp)
                temp = parent[temp]
            path.reverse()
            return path

        children = generate_children(current)
        print(f"Generating neighbors/children (Count: {len(children)}):")
        for idx, child in enumerate(children, 1):
            print(f"  Child {idx}:")
            print_state(child)

            if child in CLOSED:
                print("    -> Result: Ignored (Already fully explored in CLOSED)")
                continue

            new_g = current_g + 1
            child_h = heuristic(child, goal)
            child_f = new_g + child_h

            if child not in g or new_g < g[child]:
                g[child] = new_g
                parent[child] = current
                heapq.heappush(OPEN, (child_f, counter, child))
                counter += 1
                print(f"    -> Result: ADDED to OPEN (g={new_g}, h={child_h}, f={child_f})")
            else:
                print(f"    -> Result: Ignored (A better or equal path with g={g[child]} already exists)")
        print()
        step_num += 1

    print("--- FAILURE: Search Complete, No Solution Found ---\n")
    return None






# Initial state
start = (
    (2, 8, 3),
    (1, 6, 4),
    (7, 0, 5)
)

# Goal state
goal = (
    (1, 2, 3),
    (8, 0, 4),
    (7, 6, 5)
)
print("======================================")
print("CASE 1: MISPLACED TILES")
print("======================================")
path1 = A_STAR(start, goal, misplaced_tiles)

if path1:
    print("Solution Found!")
    print("Number of moves:", len(path1) - 1)
    print()

    for i, state in enumerate(path1):
        g_value = i
        h_value = misplaced_tiles(state, goal)
        f_value = g_value + h_value

        print("State", i)
        print_state(state)
        print("g =", g_value)
        print("h =", h_value)
        print("f =", f_value)
        print("--------------------------------")
else:
    print("No solution found.")

print("\n")
print("======================================")
print("CASE 2: MANHATTAN DISTANCE")
print("======================================")

path2 = A_STAR(start, goal, manhattan_distance)

if path2:
    print("Solution Found!")
    print("Number of moves:", len(path2) - 1)
    print()

    for i, state in enumerate(path2):
        g_value = i
        h_value = manhattan_distance(state, goal)
        f_value = g_value + h_value

        print("State", i)
        print_state(state)
        print("g =", g_value)
        print("h =", h_value)
        print("f =", f_value)
        print("--------------------------------")
else:
    print("No solution found.")
```
