1) TIC-TAC-TOE GAME

```python

board = [' ' for _ in range(9)]

# Display the board
def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):
    win_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in win_positions:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def check_draw():
    return ' ' not in board


current_player = 'X'

print("=== TIC TAC TOE ===")
print("Positions are:")
print("1 | 2 | 3")
print("--+---+--")
print("4 | 5 | 6")
print("--+---+--")
print("7 | 8 | 9")

while True:
    display_board()

    try:
        position = int(input(f"Player {current_player}, enter position (1-9): "))

        if position < 1 or position > 9:
            print("Please enter a number between 1 and 9.")
            continue

        position = position - 1

        if board[position] != ' ':
            print("Position already occupied! Try again.")
            continue

        board[position] = current_player

    except ValueError:
        print("Please enter a valid number.")
        continue

    # Check winner
    if check_winner(current_player):
        display_board()
        print(f"Player {current_player} wins!")
        break

    # Check draw
    if check_draw():
        display_board()
        print("Game is a draw!")
        break

    # Change player
    if current_player == 'X':
        current_player = 'O'
    else:
        current_player = 'X'
```

2) VACCUM CLEANER AGENT using the concept of "Simple Reflex agent and Goal based"

```python


n = int(input("Enter number of rooms: "))

rooms = []

# Input room status
for i in range(n):
    status = int(input(f"Enter status of Room {i + 1} (1=Dirty, 0=Clean): "))
    rooms.append(status)

print("\nInitial Room Status:", rooms)

# Goal: All rooms must be clean
goal = [0] * n

print("\n--- Vacuum Cleaner Agent ---")

# Visit each room
for i in range(n):

    print(f"\nVacuum is in Room {i + 1}")

    # Simple Reflex Agent
    if rooms[i] == 1:
        print("Room is Dirty -> Clean the room")
        rooms[i] = 0
    else:
        print("Room is Clean -> Move to next room")

    # Check Goal
    if rooms == goal:
        print("\nGoal Achieved: All rooms are clean!")
        break

# Final state
print("\nFinal Room Status:", rooms)

if rooms == goal:
    print(" Goal achieved successfully.")
else:
    print("Goal-Based Agent: Goal not achieved.")
```
