```python
import copy

class Node:
    """Represents a state in the 8-puzzle search space."""
    def __init__(self, board, parent=None, action=None, depth=0):
        self.board = board
        self.parent = parent
        self.action = action
        self.depth = depth

    def __eq__(self, other):
        if not isinstance(other, Node):
            return NotImplemented
        return self.board == other.board

    def __hash__(self):
        return hash(tuple(map(tuple, self.board)))

    def __str__(self):
        board_str = "\n".join([" ".join(map(str, row)) for row in self.board])
        return f"Board:\n{board_str}\nDepth: {self.depth}\nAction: {self.action}"

def find_blank(board):
    """Finds the position (row, col) of the blank space (0) in the board."""
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c
    return None

def get_neighbors(node):
    """Generates all possible next states (neighbors) from the current node."""
    neighbors = []
    r_blank, c_blank = find_blank(node.board)

    moves = {
        'up': (-1, 0),
        'down': (1, 0),
        'left': (0, -1),
        'right': (0, 1)
    }

    for action, (dr, dc) in moves.items():
        new_r, new_c = r_blank + dr, c_blank + dc

        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_board = copy.deepcopy(node.board)
            new_board[r_blank][c_blank] = new_board[new_r][new_c]
            new_board[new_r][new_c] = 0
            neighbors.append(Node(new_board, node, action, node.depth + 1))
    return neighbors

def is_goal(node, goal_board):
    """Checks if the current node's board matches the goal board."""
    return node.board == goal_board

def reconstruct_path(goal_node):
    """Reconstructs the path from the initial state to the goal state."""
    path = []
    current = goal_node
    while current:
        path.append(current)
        current = current.parent
    return path[::-1]
```

DFS

```python
def dfs_recursive(node, goal_board, visited_set, max_depth):
    """Solves the 8-Puzzle problem recursively using Depth-First Search."""
    if node in visited_set:
        return None

    visited_set.add(node)

    if is_goal(node, goal_board):
        return node

    if node.depth >= max_depth:
        return None

    for neighbor in get_neighbors(node):
        result = dfs_recursive(neighbor, goal_board, visited_set, max_depth)
        if result:
            return result

    return None
```

```python
# Example Usage for DFS
initial_board_state = [
    [2, 8, 3],
    [1, 6, 4],
    [7, 0, 5]
]

goal_board_state = [
    [1, 2, 3],
    [8, 0, 4],
    [7, 6, 5]
]

print("--- DFS Example ---")
print("Initial Board:")
for row in initial_board_state:
    print(row)

print("\nGoal Board:")
for row in goal_board_state:
    print(row)

start_node = Node(initial_board_state)
visited_states_dfs = set() # Use a distinct set for DFS

solution_node_dfs = dfs_recursive(start_node, goal_board_state, visited_states_dfs, max_depth=30)

if solution_node_dfs:
    path_dfs = reconstruct_path(solution_node_dfs)
    print(f"\nDFS Solution found in {len(path_dfs) - 1} steps (at depth {solution_node_dfs.depth}):")
    for i, node in enumerate(path_dfs):
        print(f"\nStep {i} (Action: {node.action if node.action else 'Initial State'}):")
        for row in node.board:
            print(row)
else:
    print(f"\nNo DFS solution found within the specified max_depth ({30}).")



```

IDDFS(ITERATIVE DEEPENING DLS)

```python
def dls_recursive(node, goal_board, visited_set, limit):
    """Depth-Limited Search (DLS) recursive function. Used by IDDFS."""
    if node in visited_set:
        return None

    visited_set.add(node)

    if is_goal(node, goal_board):
        return node

    if node.depth >= limit:
        return None # Depth limit reached for this DLS iteration

    for neighbor in get_neighbors(node):
        result = dls_recursive(neighbor, goal_board, visited_set, limit)
        if result:
            return result

    return None

def iddfs(initial_board, goal_board, max_depth=50):
    """Iterative Deepening Depth-First Search (IDDFS)."""
    for depth_limit in range(max_depth):
        print(f"Searching with depth limit: {depth_limit}")
        visited_set = set() # Reset visited set for each new depth limit
        start_node = Node(initial_board)
        solution = dls_recursive(start_node, goal_board, visited_set, depth_limit)
        if solution:
            return solution
    return None
```

```python
# Example Usage for IDDFS
initial_board_state_iddfs = [
    [2, 8, 3],
    [1, 6, 4],
    [7, 0, 5]
]

goal_board_state_iddfs = [
    [1, 2, 3],
    [8, 0, 4],
    [7, 6, 5]
]

print("\n--- IDDFS Example ---")
print("Initial Board:")
for row in initial_board_state_iddfs:
    print(row)

print("\nGoal Board:")
for row in goal_board_state_iddfs:
    print(row)

solution_node_iddfs = iddfs(initial_board_state_iddfs, goal_board_state_iddfs, max_depth=20) # Setting a reasonable max_depth for IDDFS

if solution_node_iddfs:
    path_iddfs = reconstruct_path(solution_node_iddfs)
    print(f"\nIDDFS Solution found in {len(path_iddfs) - 1} steps (at depth {solution_node_iddfs.depth}):")
    for i, node in enumerate(path_iddfs):
        print(f"\nStep {i} (Action: {node.action if node.action else 'Initial State'}):")
        for row in node.board:
            print(row)
else:
    print(f"\nNo IDDFS solution found within the IDDFS max_depth of {20}.")
```
