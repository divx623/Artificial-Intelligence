```python
def calculate_cost(state):
    n = len(state)
    cost = 0

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing(state):
    n = len(state)
    iteration = 0

    while True:
        current_cost = calculate_cost(state)

        print("\nIteration", iteration)
        print("Current state:", state)
        print("Cost:", current_cost)

        if current_cost == 0:
            print("\nSolution found!")
            print("Final state:", state)
            return

        best_state = state[:]
        best_cost = current_cost

        # Generate neighbors by swapping two queens
        for i in range(n):
            for j in range(i + 1, n):
                neighbor = state[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                neighbor_cost = calculate_cost(neighbor)

                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_state = neighbor

        # Stop if no better neighbor exists
        if best_cost >= current_cost:
            print("\nNo better neighbor found.")
            print("Local minimum reached.")
            print("Final state:", state)
            return

        print("Best neighbor:", best_state)
        print("Best neighbor cost:", best_cost)

        state = best_state
        iteration += 1
```

```python

n = int(input("Enter N: "))
state = list(map(int, input("Enter initial row positions: ").split()))

if n < 1 or len(state) != n:
    print("Invalid input!")
elif sorted(state) != list(range(n)):
    print("Enter a permutation of rows from 0 to N-1.")
else:
    hill_climbing(state)
```
