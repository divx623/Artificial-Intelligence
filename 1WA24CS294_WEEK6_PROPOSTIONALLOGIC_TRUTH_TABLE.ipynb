```python
import pandas as pd
import itertools
import re

def evaluate(expr, model):
    return eval(expr, {}, model)

def get_unique_symbols(kb, alpha):
    # Extract all words/symbols, ignoring Python keywords
    keywords = {'and', 'or', 'not', 'True', 'False'}
    all_exprs = kb + [alpha]
    symbols = set()
    for expr in all_exprs:
        found = re.findall(r'\b[a-zA-Z_][a-zA-Z0-9_]*\b', expr)
        symbols.update(found)
    return sorted(list(symbols - keywords))

def generate_simple_truth_table(kb, alpha):
    symbols = get_unique_symbols(kb, alpha)
    rows = []

    for truth_values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, truth_values))

        # Evaluate each clause in the KB
        kb_evals = [evaluate(clause, model) for clause in kb]
        kb_true = all(kb_evals)

        # Evaluate the query alpha
        alpha_true = evaluate(alpha, model)

        # Entailment condition: KB => alpha (which is equivalent to: not KB or alpha)
        entails = (not kb_true) or alpha_true

        row = {**model, 'KB': kb_true, 'alpha (Query)': alpha_true, 'KB => alpha': entails}
        rows.append(row)

    df = pd.DataFrame(rows)
    overall_entails = df[df['KB'] == True]['alpha (Query)'].all() if any(df['KB']) else True
    return df, overall_entails

# Demonstration with 3 variables (p, q, r)
print("=== Scenario 3 (Generalized Case with 3 variables: p, q, r) ===")
kb = ["p and q"]
query = "p"
df, result = generate_simple_truth_table(kb, query)
display(df)
print(f"Does KB {kb} entail '{query}'? {'YES' if result else 'NO'}")
```
